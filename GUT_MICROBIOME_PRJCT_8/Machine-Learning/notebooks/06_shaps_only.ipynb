```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import shap
```

```python
import json
import joblib

# Data
X_train = pd.read_csv("../data/X_train.csv")
X_test = pd.read_csv("../data/X_test.csv")

y_train = pd.read_csv("../data/y_train.csv")["Disease"]
y_test = pd.read_csv("../data/y_test.csv")["Disease"]

# Feature sets
with open("../data/feature_sets.json", "r") as f:
    feature_sets = json.load(f)
```

```python
#recreate the tuned model


from sklearn.ensemble import RandomForestClassifier

features = feature_sets["RF10"]

Xtr = X_train[features]
Xte = X_test[features]

best_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    max_features="sqrt",
    min_samples_leaf=4,
    min_samples_split=2,
    random_state=42
)

best_rf.fit(Xtr, y_train)
```

```python
#compute the shap values
explainer = shap.TreeExplainer(best_rf)

shap_values = explainer.shap_values(Xte)
```

```python
shap.summary_plot(
    shap_values[:, :, 1],
    Xte
)
```

```python
import matplotlib.pyplot as plt

shap.summary_plot(
    shap_values[:, :, 1],
    Xte,
    show=False
)
plt.title("SHAP summary")

fig = plt.gcf()
fig.savefig("../figures/shap_summary.svg",dpi=1200, bbox_inches="tight")
```

```python
#mean absolute shap values

shap.summary_plot(
    shap_values[:, :, 1],
    Xte,
    plot_type="bar"
)
```

```python
#shap dependencies plots 

shap.dependence_plot(
    "Terrisporobacter",
    shap_values[:, :, 1],
    Xte
)
```

```python
#local explanation

sample = 0

shap.plots.waterfall(
    shap.Explanation(
        values=shap_values[sample, :, 1],
        base_values=explainer.expected_value[1],
        data=Xte.iloc[sample],
        feature_names=Xte.columns
    )
)
```

```python
#save the shap values
importance = pd.DataFrame({
    "Genus": Xte.columns,
    "Mean_SHAP": np.abs(shap_values[:, :, 1]).mean(axis=0)
})

importance = importance.sort_values(
    "Mean_SHAP",
    ascending=False
)

importance.to_csv(
    "../data/shap_importance.csv",
    index=False
)
```
