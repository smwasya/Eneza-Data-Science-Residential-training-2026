```python
#load libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

```python
#load data
df = pd.read_csv("../data/machine_learning_dataset.csv")

df.head()
```

```python
#inspect data
print(df.shape)

df.info()

df.describe()
```

```python
#check missing values
df.isnull().sum().sort_values(ascending=False).head(20)
```

```python
df.isnull().sum().sum()
```

```python
#check duplicated
df["SampleID"].duplicated().sum()

df["SampleID"].nunique()
```

```python
#outcome variables
df["Disease"].value_counts()
df["Disease"].value_counts(normalize=True) * 100
```

```python
#separate feature and labels
X = df.drop(columns = ["SampleID", "Disease"])
y = df["Disease"]

X.dtypes.value_counts()
```

```python
#remove zero variance features
from sklearn.feature_selection import VarianceThreshold

selector = VarianceThreshold(threshold=0)

X = pd.DataFrame(
    selector.fit_transform(X),
    columns=X.columns[selector.get_support()]
)

print(X.shape)
```

```python
#remove near-zero prevalence genra
prevalence = (X > 0).sum(axis=0)

keep = prevalence >= 0.10 * len(X)

X = X.loc[:, keep]

print(X.shape)
```

```python
#train test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
#save processed data 
train = X_train.copy()
train["Disease"] = y_train.values

test = X_test.copy()
test["Disease"] = y_test.values

train.to_csv("../data/train.csv", index=False)
test.to_csv("../data/test.csv", index=False)
```

```python
X_filtered = X.copy()
X_filtered["Disease"] = y.values

# Include SampleID again
X_filtered.insert(0, "SampleID", df["SampleID"])

X_filtered.to_csv(
    "../data/machine_learning_dataset_filtered.csv",
    index=False
)

print(X_filtered.shape)
```
