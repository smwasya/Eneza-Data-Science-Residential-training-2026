```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_selection import VarianceThreshold
from sklearn.ensemble import RandomForestClassifier

from scipy.stats import mannwhitneyu

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/machine_learning_dataset_filtered.csv")

df.head()
```

```python
#separate features and labels
X = df.drop(columns=["SampleID", "Disease"])

y = df["Disease"]
```

```python
#Encode the target

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y = encoder.fit_transform(y)

print(encoder.classes_)
```

```python
#train/test split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(X_train.shape)
print(X_test.shape)
```

```python
#remove highly correlated genera to remove redudancy 
corr = X_train.corr().abs()

upper = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

to_drop = [
    column
    for column in upper.columns
    if any(upper[column] > 0.90)
]

print(len(to_drop))
```

```python
#remove them
X_train = X_train.drop(columns=to_drop)

X_test = X_test.drop(columns=to_drop)

print(X_train.shape)
```

```python
#mann-whitney u test
#rank genera by whether their abundance differs between health and t2d

healthy = X_train[y_train == 0]
diabetes = X_train[y_train == 1]

pvals = []

for genus in X_train.columns:

    _, p = mannwhitneyu(
        healthy[genus],
        diabetes[genus],
        alternative="two-sided"
    )

    pvals.append(p)

results = pd.DataFrame({

    "Genus": X_train.columns,

    "Pvalue": pvals

})

results = results.sort_values("Pvalue")

results.head(20)
```

```python
# multiple testing correction 
#since we are testing many genera we control fdr

from statsmodels.stats.multitest import multipletests

results["FDR"] = multipletests(
    results["Pvalue"],
    method="fdr_bh"
)[1]

results.head(20)
```

```python
#keep significant genera
significant = results[
    results["FDR"] < 0.05
]

print(significant.shape)
```

```python
#random forest feature importance 
rf = RandomForestClassifier(
    n_estimators=500,
    random_state=42
)

rf.fit(X_train, y_train)
```

```python
#extract importance
importance = pd.DataFrame({

    "Genus": X_train.columns,

    "Importance": rf.feature_importances_

})

importance = importance.sort_values(
    "Importance",
    ascending=False
)

importance.head(20)
```

```python
top20 = importance.head(20)

plt.figure(figsize=(8,7))

plt.barh(
    top20["Genus"][::-1],
    top20["Importance"][::-1]
)

plt.xlabel("Importance")

plt.title("Random Forest Feature Importance")

plt.show()
```

```python
results.to_csv(
    "../data/mann_whitney_results.csv",
    index=False
)

importance.to_csv(
    "../data/rf_feature_importance.csv",
    index=False
)
```

```python
#lasso feature selection

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
```

```python
#lasso feature selection
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("lasso", LogisticRegression(
        penalty="l1",
        solver="liblinear",
        max_iter=5000,
        random_state=42
    ))
])

param_grid = {
    "lasso__C": np.logspace(-3, 2, 20)
}

grid = GridSearchCV(
    pipe,
    param_grid,
    cv=5,
    scoring="roc_auc",
    n_jobs=-1
)

grid.fit(X_train, y_train)

print("Best C:", grid.best_params_["lasso__C"])
print("Best CV ROC-AUC:", grid.best_score_)
```

```python
best_model = grid.best_estimator_

coef = best_model.named_steps["lasso"].coef_[0]

lasso_results = pd.DataFrame({
    "Genus": X_train.columns,
    "Coefficient": coef
})

lasso_results["AbsCoefficient"] = lasso_results["Coefficient"].abs()

lasso_results = lasso_results.sort_values(
    "AbsCoefficient",
    ascending=False
)

lasso_results.head(20)
```

```python
selected = lasso_results[
    lasso_results["Coefficient"] != 0
]

print(f"Selected {len(selected)} genera")
```

```python
top = selected.head(20)

plt.figure(figsize=(8,7))

plt.barh(
    top["Genus"][::-1],
    top["Coefficient"][::-1]
)

plt.xlabel("LASSO Coefficient")
plt.title("Top LASSO Features")

plt.show()
```

```python
lasso_results.to_csv(
    "../data/lasso_feature_importance.csv",
    index=False
)
```
